# Planning

Dyna-Q learns from real transitions and repeats Q-learning updates sampled from its model. Sweeping gives priority to large TD errors.

The code imports the implementation from `rl_from_scratch` and keeps the experiment here.

```python
from rl_from_scratch import GridWorld
```

## Define the environment, algorithm, and training run

```python
from rl_from_scratch.planning import DeterministicModel, RandomSamplePlanning, DynaQ, PrioritizedSweeping
env = GridWorld(3, 3)
agent = DynaQ(env, n_planning=5, alpha=0.2, gamma=0.9, epsilon=0.1, seed=7)
result = agent.train(200, max_steps=100)
```

## Inspect the learned result

```python
print('Observed model entries:', len(agent.model.experience))
print('Greedy first action:', result.policy[env.start])
```

## Visualize learning

```python
import matplotlib.pyplot as plt
plt.plot(result.episode_lengths); plt.xlabel('Episode'); plt.ylabel('Steps'); plt.title('Dyna-Q planning'); plt.show()
```
