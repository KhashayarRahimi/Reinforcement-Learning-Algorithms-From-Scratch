# Eligibility Traces

Eligibility traces assign a TD error to recently active features. Dutch traces add a correction for true-online learning.

The code imports the implementation from `rl_from_scratch` and keeps the experiment here.

```python
from rl_from_scratch import GridWorld
```

## Define the environment, algorithm, and training run

```python
from rl_from_scratch.eligibility_traces import SemiGradientTDLambda, TrueOnlineTDLambda, SarsaLambda, TrueOnlineSarsaLambda
from rl_from_scratch.value_functions import GridFeatures, LinearValue
from rl_from_scratch.policies import TabularPolicy
env = GridWorld(3, 3)
policy = TabularPolicy({s: {a: 0.25 for a in env.actions(s)} for s in env.states if env.actions(s)})
value = LinearValue(GridFeatures(env.rows, env.cols, env.goal), dimension=3)
agent = TrueOnlineTDLambda(env, policy, value, alpha=0.01, gamma=0.9, lam=0.8, seed=7)
result = agent.train(300, max_steps=100)
```

## Inspect the learned result

```python
print('Weights:', result.parameters)
print('Start value:', result.values[env.start])
```

## Visualize learning

```python
import matplotlib.pyplot as plt
plt.plot(result.episode_returns); plt.xlabel('Episode'); plt.ylabel('Return'); plt.title('True online TD(lambda)'); plt.show()
```
