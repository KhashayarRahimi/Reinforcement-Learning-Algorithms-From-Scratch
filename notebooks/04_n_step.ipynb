# N Step

The n-step target sums n discounted rewards, then bootstraps. Tree Backup and Q(σ) mix expected and sampled values.

The code imports the implementation from `rl_from_scratch` and keeps the experiment here.

```python
from rl_from_scratch import GridWorld
```

## Define the environment, algorithm, and training run

```python
from rl_from_scratch.n_step import NStepTD, NStepSarsa, OffPolicyNStepSarsa, TreeBackup, QSigma
env = GridWorld(3, 3)
agent = NStepSarsa(env, n=3, alpha=0.2, gamma=0.9, epsilon=0.1, seed=7)
result = agent.train(300, max_steps=100)
```

## Inspect the learned result

```python
print('Greedy first action:', result.policy[env.start])
print('Final ten returns:', result.episode_returns[-10:])
```

## Visualize learning

```python
import matplotlib.pyplot as plt
plt.plot(result.episode_returns); plt.xlabel('Episode'); plt.ylabel('Return'); plt.title('Three-step Sarsa'); plt.show()
```
