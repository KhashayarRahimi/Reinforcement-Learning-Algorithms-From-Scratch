# Monte Carlo

First-visit MC averages complete returns. Epsilon-soft control improves a sampled action-value estimate.

The code imports the implementation from `rl_from_scratch` and keeps the experiment here.

```python
from rl_from_scratch import GridWorld
```

## Define the environment, algorithm, and training run

```python
from rl_from_scratch.monte_carlo import FirstVisitPrediction, OnPolicyControl, OffPolicyPrediction, OffPolicyControl
from rl_from_scratch.policies import TabularPolicy
env = GridWorld(3, 3)
random_policy = TabularPolicy({s: {a: 0.25 for a in env.actions(s)} for s in env.states if env.actions(s)})
agent = OnPolicyControl(env, gamma=0.9, epsilon=0.1, seed=7)
result = agent.train(300, max_steps=100)
```

## Inspect the learned result

```python
print('Greedy first action:', result.policy[env.start])
print('First ten returns:', result.episode_returns[:10])
```

## Visualize learning

```python
import matplotlib.pyplot as plt
plt.plot(result.episode_returns, alpha=0.5); plt.xlabel('Episode'); plt.ylabel('Return'); plt.title('Monte Carlo control'); plt.show()
```
