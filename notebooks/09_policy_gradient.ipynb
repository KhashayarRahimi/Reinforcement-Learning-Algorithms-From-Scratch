# Policy Gradient

REINFORCE weights the score gradient by a sampled return. Actor–critic replaces that return with a TD error.

The code imports the implementation from `rl_from_scratch` and keeps the experiment here.

```python
from rl_from_scratch import GridWorld
```

## Define the environment, algorithm, and training run

```python
from rl_from_scratch.policy_gradient import REINFORCE, REINFORCEWithBaseline, OneStepActorCritic, EpisodicTraceActorCritic, ContinuingTraceActorCritic
from rl_from_scratch.policies import LinearSoftmaxPolicy
from rl_from_scratch.value_functions import GridFeatures, LinearValue
env = GridWorld(3, 3)
base = GridFeatures(env.rows, env.cols, env.goal)
actions = tuple(env.actions(env.start))
def actor_features(state, action):
    import numpy as np
    x = np.zeros(3*len(actions))
    x[actions.index(action)*3:(actions.index(action)+1)*3] = base(state)
    return x
actor = LinearSoftmaxPolicy(actor_features, dimension=3*len(actions))
critic = LinearValue(base, dimension=3)
agent = REINFORCEWithBaseline(env, actor, critic=critic, actor_alpha=0.001, critic_alpha=0.01, gamma=0.9, seed=7)
result = agent.train(300, max_steps=100)
```

## Inspect the learned result

```python
print('Greedy first action:', result.policy[env.start])
print('Critic start value:', result.values[env.start])
```

## Visualize learning

```python
import matplotlib.pyplot as plt
plt.plot(result.episode_returns); plt.xlabel('Episode'); plt.ylabel('Return'); plt.title('REINFORCE with baseline'); plt.show()
```
