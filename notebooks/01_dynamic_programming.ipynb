# Dynamic Programming

Policy evaluation computes a Bellman expectation; policy iteration alternates evaluation and greedy improvement.

The code imports the implementation from `rl_from_scratch` and keeps the experiment here.

```python
from rl_from_scratch import GridWorld
```

## Define the environment, algorithm, and training run

```python
from rl_from_scratch.dynamic_programming import PolicyEvaluation, PolicyIteration
from rl_from_scratch.policies import TabularPolicy
env = GridWorld(3, 3, goal=(2, 2))
random_policy = TabularPolicy({s: {a: 0.25 for a in env.actions(s)} for s in env.states if env.actions(s)})
agent = PolicyEvaluation(env, random_policy, gamma=0.9)
result = agent.run()
optimal = PolicyIteration(env, gamma=0.9).run()
```

## Inspect the learned result

```python
print('Random-policy value:', result.values[env.start])
print('Improved first action:', optimal.policy[env.start])
```

## Visualize learning

```python
import numpy as np
import matplotlib.pyplot as plt
values = np.array([[optimal.values[r, c] for c in range(env.cols)] for r in range(env.rows)])
plt.imshow(values); plt.colorbar(label='Optimal state value'); plt.title('Policy iteration'); plt.show()
```
