# Approximation Control

Semi-gradient Sarsa updates a linear action-value function using the sampled next action.

The code imports the implementation from `rl_from_scratch` and keeps the experiment here.

```python
from rl_from_scratch import GridWorld
```

## Define the environment, algorithm, and training run

```python
from rl_from_scratch.function_approximation import SemiGradientSarsa, NStepSemiGradientSarsa, DifferentialSarsa, DifferentialNStepSarsa
from rl_from_scratch.value_functions import GridFeatures, LinearActionValue
env = GridWorld(3, 3)
base = GridFeatures(env.rows, env.cols, env.goal)
actions = tuple(env.actions(env.start))
def features(state, action):
    import numpy as np
    x = np.zeros(3 * len(actions))
    x[actions.index(action)*3:(actions.index(action)+1)*3] = base(state)
    return x
value = LinearActionValue(features, dimension=3*len(actions))
agent = SemiGradientSarsa(env, value, alpha=0.01, gamma=0.9, epsilon=0.1, seed=7)
result = agent.train(300, max_steps=100)
```

## Inspect the learned result

```python
print('Greedy first action:', result.policy[env.start])
print('Weight norm:', sum(result.parameters ** 2) ** 0.5)
```

## Visualize learning

```python
import matplotlib.pyplot as plt
plt.plot(result.episode_returns); plt.xlabel('Episode'); plt.ylabel('Return'); plt.title('Linear Sarsa control'); plt.show()
```
