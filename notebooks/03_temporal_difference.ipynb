# Temporal Difference

Q-learning bootstraps from the best successor action; Sarsa bootstraps from the action actually selected.

The code imports the implementation from `rl_from_scratch` and keeps the experiment here.

```python
from rl_from_scratch import GridWorld
```

## Define the environment, algorithm, and training run

```python
from rl_from_scratch.temporal_difference import TDZero, Sarsa, QLearning, DoubleQLearning
env = GridWorld(3, 3)
agent = QLearning(env, alpha=0.2, gamma=0.9, epsilon=0.1, seed=7)
result = agent.train(300, max_steps=100)
```

## Inspect the learned result

```python
print('Q at start:', {a: result.action_values.get((env.start, a), 0) for a in env.actions(env.start)})
print('Greedy first action:', result.policy[env.start])
```

## Visualize learning

```python
import matplotlib.pyplot as plt
plt.plot(result.episode_returns); plt.xlabel('Episode'); plt.ylabel('Return'); plt.title('Q-learning'); plt.show()
```
